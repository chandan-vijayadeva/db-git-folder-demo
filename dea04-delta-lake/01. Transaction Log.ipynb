## 0. Create a new schema under the demo catalog for this section of course (delta_lake)

```python
%sql
CREATE CATALOG IF NOT EXISTS demo
      MANAGED LOCATION 'abfss://demo@deacourseexternaldl.dfs.core.windows.net' 
      COMMENT 'This is the catalog for the demo Data Lakehouse'
```

```python
%sql
create schema if not exists demo.delta_lake
      MANAGED LOCATION 'abfss://demo@deacourseexternaldl.dfs.core.windows.net/delta_lake' 
      COMMENT 'This is the schema for the demo Data Lakehouse'
```

## 1. Create a Delta Lake Table

```python
%sql
create table if not exists demo.delta_lake.companies
(company_name string,
founded_date Date,
country string);
```
