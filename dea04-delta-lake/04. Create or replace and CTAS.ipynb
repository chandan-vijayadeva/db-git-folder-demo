## Create ot Replace and CTAS
- Difference betweeen Create or replace and drop and create table statemtnets
- CTAS statemtne

```sql
-- 1. behavior of the drop and create statement

DROP table if exists demo.delta_lake.companies;

create table demo.delta_lake.companies
(
    company_id bigint not null generated always as identity (start with 1 increment by 1),
    company_name STRING NOT null,
    founded_date DATE comment 'This date the company was founded',
    country STRING 
)
comment 'This table contains information about some of the successful tech companies'
tblproperties ('sensitive' = 'true', 'delta.enableDeletionVectors' = 'false');


insert into demo.delta_lake.companies
(company_name, founded_date, country)
values ( "Apple", "1970-04-01", "USA"),
        ("Microsoft", "1975-04-04", "USA"),
        ("Google", "1998-09-04", "USA"),
        ("Amazon","1994-07-05","USA")
```

```sql
desc history demo.delta_lake.companies;
```

```sql

-- Behavior of the create or replace statement

DROP table if exists demo.delta_lake.companies;

```

```sql
desc history demo.delta_lake.companies
```

```sql
-- Behavior of the create or replace statement

create or replace table demo.delta_lake.companies
(
    company_id bigint not null generated always as identity (start with 1 increment by 1),
    company_name STRING NOT null,
    founded_date DATE comment 'This date the company was founded',
    country STRING 
)
comment 'This table contains information about some of the successful tech companies'
tblproperties ('sensitive' = 'true', 'delta.enableDeletionVectors' = 'false');


insert into demo.delta_lake.companies
(company_name, founded_date, country)
values ( "Apple", "1970-04-01", "USA"),
        ("Microsoft", "1975-04-04", "USA"),
        ("Google", "1998-09-04", "USA"),
        ("Amazon","1994-07-05","USA")
```

```sql
desc history demo.delta_lake.companies
```

## 2. CTAS statement (Create table as select statement)

```sql

DROP table demo.delta_lake.companies_china;
```

```sql
create table demo.delta_lake.companies_china
as
select CAST(company_id as INT) as company_id,
        company_name,
        founded_date,
        country
 from demo.delta_lake.companies
where country = 'China';
```

```sql
desc extended demo.delta_lake.companies_china
```

```sql
select * from demo.delta_lake.companies_china
```

```sql
-- if I want to add comment for founded date, we need to alter table, as it's infered from other table

ALTER table demo.delta_lake.companies_china
alter column founded_date COMMENT 'Date comapny was found'
```

```sql
-- if we want to have not null constraints
alter table demo.delta_lake.companies_china
alter column company_id set not null;
```

```sql
desc history demo.delta_lake.companies_china
```
