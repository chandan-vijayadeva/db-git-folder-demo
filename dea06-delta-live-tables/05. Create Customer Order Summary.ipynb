## Create Customer Order Summary
- Join the table silcer_customers, silver_addresses and silver_orders
- Retreive latest address of the customer
- Calculate the following values
    - total_orders
    - total_items_ordered
    - total_order_amount

```sql
create or refresh materialized view gold_customer_order_summary
as
select c.customer_id,
       c.customer_name,
       c.date_of_birth,
       c.telephone,
       c.email,
       a.address_line_1,
       a.city,
       a.state,
       a.postcode,
       COUNT(DISTINCT o.order_id) as total_orders,
       SUM(o.item_quantity) as total_items_orderd,
       SUM(o.item_quantity * o.item_price) as total_ordered_amount 
from LIVE.silver_customer c 
join LIVE.silver_addresses a on c.customer_id = a.customer_id
join LIVE.silver_orders o on c.customer_id = o.customer_id
where a.__END_AT is null
group by all;
```
