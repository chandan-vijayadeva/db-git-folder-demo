## Set-up the project enviornment for citrusbox data lakehouse
- Create external location - dea_course_ext_dl_circuitbox
- Create Catalog - circuitbox
- Create Schemas
    - landing
    - lakehouse
- Create Volume - operational_data

```sql
-- 1 Create External Location

create external location if not exists dea_course_ext_dl_circuitbox
    url 'abfss://circuitbox@deacourseexternaldl.dfs.core.windows.net/' --adls path
    with (storage credential dea_course_ext_sc)
    comment 'External location for the circuitbox data lakehouse'
```

```sql
%fs ls 'abfss://circuitbox@deacourseexternaldl.dfs.core.windows.net/'
```

```sql
-- 2 Create catalog

create catalog if not exists circuitbox
    managed location 'abfss://circuitbox@deacourseexternaldl.dfs.core.windows.net/'
    comment 'Catalog for the circuitbox data lakehouse'
```

```sql
SHOW CATALOGS
```

```sql
-- 3 Create Schemas

use catalog circuitbox;

create schema if not exists landing
    managed location 'abfss://circuitbox@deacourseexternaldl.dfs.core.windows.net/landing';

create schema if not exists lakehouse
    managed location 'abfss://circuitbox@deacourseexternaldl.dfs.core.windows.net/lakehouse';

```

```sql
show schemas;
```

```sql
--4 Create volume

use catalog circuitbox;

use schema landing;

create external volume if not exists operational_data
    location 'abfss://circuitbox@deacourseexternaldl.dfs.core.windows.net/landing/operational_data';
```

```sql
%fs ls /Volumes/circuitbox/landing/operational_data
```
