## Transform membership data
1. Extract customer_id from file path
2. Write transformed data to the silver schema

```python
df_membership = spark.table('gizmobox.bronze.py_memberships')
display(df_membership)
```

## 1. Extract customerid form the file path

```python
from pyspark.sql import functions as F

df_extracted = (

                df_membership
                .select(
                    F.regexp_extract("path", r".*/([0-9]+)\.png$", 1).alias("customer_id"),
                    F.col("content").alias("membership_card")               
                    )
)

display(df_extracted)
```

## 2. Write transformed data to the silver schema

```python
df_extracted.writeTo('gizmobox.silver.py_membership').createOrReplace()
```
