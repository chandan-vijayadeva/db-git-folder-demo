
## Extract data from payments files
1. Read Payment File
2. Create Payments Table in Bronze Schema

### 1. List the files from Payment folder

```python
%fs ls 'abfss://gizmobox@deacourseexternaldl.dfs.core.windows.net/landing/external_data/payments'
```

### 2. Read the payment file

```python
## DDL Formate of creating schema
payments_schema = 'payment_id INTEGER, order_id INTEGER, payment_timestamp TIMESTAMP, payment_status INTEGER, payment_method STRING' ## DDL format
```

```python
## Python formate of creating schema
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, TimestampType

py_payments_schema = StructType([
    StructField('payment_id', IntegerType()),
    StructField('order_id', IntegerType()),
    StructField('payment_timestamp', TimestampType()),
    StructField('payment_status', IntegerType()),
    StructField('payment_method',StringType())
])
```

```python
df = (
    spark.read.format('csv')
    .option('delimiter', ',')
    .schema(py_payments_schema)
    .load('abfss://gizmobox@deacourseexternaldl.dfs.core.windows.net/landing/external_data/payments')
)

display(df)
```

## 3. Create Table in Bronze schema

```python
df.writeTo('gizmobox.bronze.py_payments').createOrReplace()
```
