
## Extract Data from the refunds sql table
1. Create Bronze schema in Hive Metastore
2. Create External Table

### 1. Read refunds data via jdbc

```python
df = (
        spark.read.format('jdbc')
        .option('url', 'jdbc:sqlserver://gizmobox-sql-srv.database.windows.net:1433;database=gizmobox-db')
        .option('dbtable','refunds')
        .option('user', 'gizmoboxadmin')
        .option('password', 'Gizmobox1')
        .load()
)

display(df)
```

## 2. Write refunds data to the bronze layer

```python
df.writeTo('gizmobox.bronze.py_refunds').createOrReplace()
```
