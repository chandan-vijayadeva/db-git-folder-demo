## 4. Select File Meatadata

```sql
SELECT input_file_name() as file_path,
*
FROM json.`/Volumes/gizmobox/landing/operational_data/customers`

-- input_file_name is depricated from databricks runtime 13.3 LTS onwards

-- there is alternative approch to use hidden colum (_metadata)
```

```sql
SELECT input_file_name() as file_path, -- input_file_name is depricated from databricks runtime 13.3 LTS onwards,  there is alternative approch to use hidden colum (_metadata)
_metadata.file_path as file_path,  -- or we can directly use _metadata
*
FROM json.`/Volumes/gizmobox/landing/operational_data/customers`
```

5.  Register Files in Unity catalog using Views

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers -- catalogname.schema name.view_name
AS 
SELECT *, 
_metadata.file_path as file_path
FROM json.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
--or we can create view in this manner as well
USE CATALOG gizmobox;
USE SCHEMA bronze;

CREATE OR REPLACE VIEW v_customers
AS
SELECT *,
_metadata.file_path as file_path
FROM json.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
SELECT * FROM gizmobox.bronze.v_customers;
```

6. Create Temporary view

```sql
--when session ends, it will be removed until it run again in next session
CREATE OR REPLACE TEMPORARY VIEW tv_customers
AS
select *,
_metadata.file_path as file_path
from json.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
select * from tv_customers;
```

7. Create Global Temporary View

```sql
-- it will work until cluster get terminated
--advantage is that we can create in this notebook and can access in other notebook of same cluster!

CREATE OR REPLACE GLOBAL TEMPORARY VIEW gtv_customers
AS
SELECT
*,
_metadata.file_path as file_path
FROM json.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
select * from global_temp.gtv_customers;  -- we should use global_temp schema to get this data of view
```
