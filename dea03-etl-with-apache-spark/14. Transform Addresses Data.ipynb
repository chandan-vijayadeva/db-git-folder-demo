
## Transform Addresses data
1. Create one record for each customer with 2 sets of address columns, 1 for shipping, 1 for billing address
2. Write transfomed data to the silver schema

```sql
select customer_id,
        address_type,
        address_line_1,
        city,
        state,
        postcode
from gizmobox.bronze.v_address
```

```sql
-- create one record for each customer with both addresses, one for each address type using PIVOT (goint form rows to column) clause

select *
from (
    select customer_id,
        address_type,
        address_line_1,
        city,
        state,
        postcode
from gizmobox.bronze.v_address
)
pivot (
    max(address_line_1) as address_line_1,
    max(city) as city,
    max(state) as state,
    max(postcode) as postcode
    for address_type in ('shipping', 'billing')
);
```

```sql
--2 write data to silver layer

create table gizmobox.silver.addresses
as
select * 
from (
    select customer_id,
    address_type,
    address_line_1,
    city,
    state,
    postcode
    from gizmobox.bronze.v_address
)
pivot (
    max(address_line_1) as address_line_1,
    max(city) as city,
    max(state) as state,
    max(postcode) as postcode
    for address_type in ('shipping', 'billing')
)
```
